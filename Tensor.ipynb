# TENSORS

*   Um tensor do PyTorch é uma matriz multidimensional que contém elementos de um único tipo de dado.

*   É semelhante aos arrays do NumPy, mas com recursos adicionais que permitem seu uso eficiente em GPUs (além das CPUs).
*   O tipo de dado padrão é **float32**.
*   É mais adequado para aprendizado profundo (deep learning) do que um array do NumPy, pois integra suporte a autograd, GPU e operações otimizadas.
